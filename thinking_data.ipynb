```python
import json
```

```python
#Load data
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
        return data 
```

```python
data = load_data("store_data.json")
print(data)
print(type(data))
# data = list
# indexes of list = customer's / user's value => dictionary
```

```python
# we know that the data is not clean :
# data is inconsitence => there are not same datatype of a single variable
# there are also some missing data present there => age 
# duplicate data is also there => like there is two alice name 

# so we have to clean & structured the data 
# do it by step by step 
```

```python
# now we write a function to clean and structure the data
# first try to print all the users by our function line by line
 # clean rating data 
        #two prolems are there :
        #(i) text & number both present => rating should be in number format
        #(ii) there are some trailing spaces => remove that trailing spaces => .strip()

def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()

    
    for user in data:
        # print(user)

        #clean rating 
        raw_rating = str(user["rating"]).strip().lower()
        # print(raw_rating)
        if(raw_rating in text_to_num):
            raw_rating = text_to_num[raw_rating]

        user["rating"] = raw_rating

        #Handling missing value
        raw_age = user.get("age")
        if(raw_age == None):
            user["age"] = None

        #Deduplication
        if (user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data

        
            
        
```

```python
data = clean_data(data)
```

```python
#Get meaningful Insights from data 
def get_insights(data):

    #avg rating 
    total_rating = 0
    for user in data:
        total_rating += float(user["rating"])
    print(f"avg rating = {total_rating / len(data)}")

    #percentage of users with poor rating
    poor_rating = 0
    for user in data:
        if(float(user["rating"]) < 3):
            poor_rating += 1
    print(f"% of users with poor rating = {poor_rating / len(data) * 100}%")
```

```python
 get_insights(data)
```

```python
# Build Recommendation feature 
def get_recommendation(data):
    recommendations = []

    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]
        curr_recomm["rating"] = user["rating"]
        
        if(float(user["rating"]) >= 4):
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"

        #store in recommendation
        recommendations.append(curr_recomm)
    return recommendations
```

```python
get_recommendation(data)
```
